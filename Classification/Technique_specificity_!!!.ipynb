# Task
Create a Python script to process "sequences_with_group_labels.csv" and "technique_hierarchy.csv". The script should read "sequences_with_group_labels.csv" row by row, extract the `intrusion_set_id` and `sequence_tokens`. For each `sequence_tokens`, replace any sub-technique with its parent technique based on the mapping provided in "technique_hierarchy.csv". Finally, create a new CSV file containing the `intrusion_set_id`, `intrusion_set_name`, and the modified `sequence_tokens`. The script should be executable in Google Colab.

## Load the datasets

### Subtask:
Load `technique_hierarchy.csv` and `sequences_with_group_labels.csv` into pandas DataFrames.


**Reasoning**:
Load the specified CSV files into pandas DataFrames.



```python
technique_hierarchy_df = pd.read_csv('/content/data/technique_hierarchy.csv')
sequences_df = pd.read_csv('/content/data/sequences_with_group_labels.csv')
```

## Create a subtechnique-to-parent technique mapping

### Subtask:
Create a subtechnique-to-parent technique mapping.


**Reasoning**:
Create an empty dictionary and iterate through the `technique_hierarchy_df` to populate it with the subtechnique-to-parent technique mapping.



```python
subtechnique_to_parent_mapping = {}
for index, row in technique_hierarchy_df.iterrows():
    if pd.notnull(row['subtechnique_id']):
        subtechnique_to_parent_mapping[row['subtechnique_id']] = row['technique_id']

# Display a few entries to verify
print("Sample subtechnique to parent mapping:")
for i, (subtech, parent_tech) in enumerate(subtechnique_to_parent_mapping.items()):
    if i < 10:
        print(f"{subtech}: {parent_tech}")
    else:
        break
```

## Process sequences

### Subtask:
Process sequences


**Reasoning**:
Iterate through the sequences DataFrame, process each sequence to replace sub-techniques with parent techniques



```python
#Reasoning: Iterate through the sequences DataFrame, process each sequence to replace sub-techniques with parent technique
import ast

processed_sequences = []

for index, row in sequences_df.iterrows():
    intrusion_set_id = row['intrusion_set_id']
    intrusion_set_name = row['intrusion_set_name']
    # Safely evaluate the string representation of the list
    sequence_tokens = ast.literal_eval(row['sequence_tokens'])

    modified_tokens = []
    for token in sequence_tokens:
        if token in subtechnique_to_parent_mapping:
            modified_tokens.append(subtechnique_to_parent_mapping[token])
        else:
            modified_tokens.append(token)

    # Append the processed sequence as a list in a single column
    processed_sequences.append({
        'intrusion_set_id': intrusion_set_id,
        'intrusion_set_name': intrusion_set_name,
        'processed_sequence_tokens': modified_tokens  # Keep the tokens as a list
    })

# Display the first few processed sequences to verify
print("Sample processed sequences:")
for i, seq_data in enumerate(processed_sequences):
    if i < 5:
        print(seq_data)
    else:
        break
```

```python
processed_sequences_df = pd.DataFrame(processed_sequences)

# Display the first few rows of the new DataFrame
print("\nProcessed sequences DataFrame:")
display(processed_sequences_df.head())
```

## Calculate non-unique technique counts

### Subtask:
Calculate the total count of each technique across all sequences, including repeated occurrences within a single sequence, and their corresponding specificity scores.


**Reasoning**:
Flatten the list of lists of technique tokens, calculate the frequency of each token, and calculate the specificity based on the frequency.



```python
from collections import Counter

# Flatten the list of lists in the 'processed_sequence_tokens' column
all_techniques = [token for sequence in processed_sequences_df['processed_sequence_tokens'] for token in sequence]

# Calculate the frequency of each technique (Non-Unique_total_count)
technique_counts = Counter(all_techniques)

# Create a pandas Series from the frequency counts
technique_counts_series = pd.Series(technique_counts, name='Non-Unique_total_count')

# Calculate the Non-Unique_specificity_count_per_technique (1 / count)
non_unique_specificity = 1 / technique_counts_series

# Create a DataFrame for non-unique counts and specificity
non_unique_analysis_df = pd.DataFrame({
    'technique_name': technique_counts_series.index,
    'non_unique_total_count': technique_counts_series.values,
    'non-unique_specificity_count_per_technique': non_unique_specificity.values
})

# Display the first few rows of the non-unique analysis DataFrame
print("non-unique technique analysis:")
display(non_unique_analysis_df.head())
```

## Calculate unique technique counts

### Subtask:
For each sequence, determine the unique techniques present and calculate the total count of each unique technique across all sequences, along with their corresponding specificity scores.


**Reasoning**:
Determine the unique techniques per sequence, flatten them, count occurrences, and calculate specificity.



```python
# Create a list of unique techniques for each sequence
unique_techniques_per_sequence = [list(set(sequence)) for sequence in processed_sequences_df['processed_sequence_tokens']]

# Flatten the list of lists of unique techniques
all_unique_techniques = [token for unique_sequence in unique_techniques_per_sequence for token in unique_sequence]

# Calculate the frequency of each unique technique (Unique_total_count)
unique_technique_counts = Counter(all_unique_techniques)

# Create a pandas Series from the unique technique counts
unique_technique_counts_series = pd.Series(unique_technique_counts, name='Unique_total_count')

# Calculate the Unique_specificity_count_per_technique (1 / count)
unique_specificity = 1 / unique_technique_counts_series

# Display the first few entries of the unique technique counts and specificity
print("Sample unique technique counts:")
display(unique_technique_counts_series.head())

print("\nSample unique specificity scores:")
display(unique_specificity.head())
```

**Reasoning**:
Combine the unique technique counts and specificity scores into a DataFrame.



```python
# Create a DataFrame for unique counts and specificity
unique_analysis_df = pd.DataFrame({
    'technique_name': unique_technique_counts_series.index,
    'unique_total_count': unique_technique_counts_series.values,
    'unique_specificity_count_per_technique': unique_specificity.values
})

# Display the first few rows of the unique analysis DataFrame
print("\n unique technique analysis:")
display(unique_analysis_df.head())
```

## Create analytical dataframe

### Subtask:
Create a new DataFrame that combines the non-unique and unique technique count and specificity analysis results.


**Reasoning**:
Merge the non-unique and unique technique analysis DataFrames on the 'تکنیک' column and display the head of the resulting DataFrame.



```python
# Merge the non-unique and unique analysis DataFrames
analytical_df = pd.merge(non_unique_analysis_df, unique_analysis_df, on='technique_name', how='outer')

# Display the first few rows of the combined DataFrame
print("Combined technique analysis DataFrame:")
display(analytical_df.head())
```

```python
import pandas as pd
import ast

# مرحله 1: خواندن فایل‌ها
processed_df = pd.read_excel('processed_sequences_table.xlsx')
technique_analysis_df = pd.read_csv('technique_analysis.csv')

# مرحله 2: ساخت دیکشنری mapping از technique_id به unique_specificity_count_per_technique
# فقط ردیف‌هایی که technique_name معتبر دارند
valid_techniques = technique_analysis_df.dropna(subset=['technique_name'])
spec_dict = dict(zip(valid_techniques['technique_name'], valid_techniques['unique_specificity_count_per_technique']))

# مرحله 3: تابع برای پردازش هر ردیف
def compute_total_specificity(token_list_str):
    try:
        tokens = ast.literal_eval(token_list_str)
    except:
        tokens = []
    # حذف تکرارها و نگهداری ترتیب
    seen = set()
    unique_tokens = []
    for t in tokens:
        if t not in seen:
            unique_tokens.append(t)
            seen.add(t)
    # محاسبه مجموع specificity
    total = 0.0
    for t in unique_tokens:
        total += spec_dict.get(t, 0.0)  # اگر تکنیک وجود نداشت، 0 اضافه شود
    return unique_tokens, total

# مرحله 4: اعمال تابع روی هر ردیف
processed_df[['processed_sequence_unique_tokens', 'total_specificity']] = processed_df['processed_sequence_tokens'].apply(
    lambda x: pd.Series(compute_total_specificity(x))
)

# مرحله 5: انتخاب ستون‌های مورد نیاز
result_df = processed_df[[
    'intrusion_set_id',
    'intrusion_set_name',
    'processed_sequence_unique_tokens',
    'total_specificity'
]]

# نمایش نتیجه
print(result_df.head())

# ذخیره به عنوان فایل CSV (اختیاری)
result_df.to_csv('apt_total_specificity.csv', index=False)
```
